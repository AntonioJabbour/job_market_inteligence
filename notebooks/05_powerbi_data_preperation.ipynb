# 5.1 Load Processed Dataset

```python
import pandas as pd

df = pd.read_csv('processed_job_data.csv')
df.head()
```

```python
print(df.columns)
print(df.shape)
```

```python
import ast 

df['extracted_skills'] = df['extracted_skills'].apply(ast.literal_eval)
df['extracted_tools'] = df['extracted_tools'].apply(ast.literal_eval)
df['extracted_technologies'] = df['extracted_technologies'].apply(ast.literal_eval)
```

```python
df['job_id'] = range(1, len(df) + 1)
```

```python
print(df['job_id'].is_unique)
print(df['job_id'].duplicated().sum())
```

```python
excluded_columns = ['extracted_skills', 'extracted_tools', 'extracted_technologies']

jobs_table = df.drop(columns=excluded_columns)
jobs_table.head()
```

```python
print(jobs_table.columns)
print(jobs_table.shape)
```

```python
extracted_columns = ['clean_job_title', 'company_name',
       'job_country', 'employment_type', 'is_remote', 'job_description',
       'job_source', 'posted_at_utc', 'normalized_job_description', 'extracted_tools', 'extracted_technologies']
skills_table = df.drop(columns= extracted_columns)

skills_table= skills_table.explode('extracted_skills')
skills_table = skills_table.rename(columns=  {'extracted_skills': 'skill'})

skills_table.head()
```

```python
print(skills_table.columns)
print(skills_table.shape)
```

```python
skills_table['skill'].isnull().sum()

skills_table = skills_table.dropna(subset=['skill'])
skills_table.shape
```

```python
skills_table.head()
```

```python
extracted_columns = ['clean_job_title', 'company_name',
       'job_country', 'employment_type', 'is_remote', 'job_description',
       'job_source', 'posted_at_utc', 'normalized_job_description', 'extracted_skills', 'extracted_technologies']
tools_table = df.drop(columns= extracted_columns)

tools_table= tools_table.explode('extracted_tools')
tools_table = tools_table.rename(columns=  {'extracted_tools': 'tool'})

tools_table.head()
```

```python
tools_table['tool'].isnull().sum()

tools_table = tools_table.dropna(subset=['tool'])
tools_table.shape
```

```python
extracted_columns = ['clean_job_title', 'company_name',
       'job_country', 'employment_type', 'is_remote', 'job_description',
       'job_source', 'posted_at_utc', 'normalized_job_description', 'extracted_skills', 'extracted_tools']
technologies_table = df.drop(columns= extracted_columns)

technologies_table= technologies_table.explode('extracted_technologies')
technologies_table = technologies_table.rename(columns=  {'extracted_technologies': 'technology'})

technologies_table.head()
```

```python
tools_table['tool'].isnull().sum()

tools_table = tools_table.dropna(subset=['tool'])
tools_table.shape
```

```python
jobs_table.to_csv('jobs_table.csv', index=False)
skills_table.to_csv('skills_table.csv', index=False)
tools_table.to_csv('tools_table.csv', index=False)
technologies_table.to_csv('technologies_table.csv', index=False)
```
