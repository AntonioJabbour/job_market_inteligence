## Raw Data Inspection — Job Market Intelligence Project

This notebook explores the raw dataset stored in the `raw_job_postings` table.  
The goal is to assess data quality issues before building the data cleaning and preprocessing pipeline.

---

### Data Source

The data is collected through the ingestion pipeline and stored in a PostgreSQL database.

- Table name: `raw_job_postings`  
- Data type: Raw, unprocessed job postings  
- This dataset includes job metadata such as title, company, location, salary, and job description.

---

### Dataset Overview

- Number of rows: 10
- Number of columns: 21

The dataset contains job postings with fields related to job details, company information, salary, and descriptions.

Initial observation:  
- The schema matches the ingestion design  
- Some columns may require type validation (especially salary and date fields)

---

### Sample Data Inspection

A sample of the dataset shows that the data is generally readable but still raw.

Observations:  
- Job titles are present but may vary in format  
- Company names are mostly available  
- Location fields appear inconsistent in some cases  
- Job descriptions are present but may contain noise or formatting issues  

---

### Missing Values Analysis

The dataset contains missing values across multiple columns.

Key observations:  
- Salary fields (`salary_min`, `salary_max`) are missing in a large portion of the dataset  
- Some job postings do not include complete location information  
- Company-related fields such as website may be sparsely populated  
- Job descriptions are mostly present but may still contain empty or low-quality entries  

Missing data will need to be handled carefully depending on its importance for analysis and modeling.

---

### Duplicate Analysis

Duplicate job postings may exist in the dataset.

Observations:  
- Some records may share the same `source_job_id`  
- There may be repeated job postings with slight variations  

Duplicate handling will be required to avoid bias in analysis and modeling.

---

### Categorical Columns Inspection

Several categorical fields show inconsistencies in formatting and values.

Observations:  
- `employment_type` contains inconsistent labels (e.g., FULL_TIME vs Full-time)  
- `job_country` and `job_state` are not fully standardized  
- `is_remote` may have inconsistent or missing values  

These fields will require normalization to ensure consistency.

---

### Salary Analysis

Salary data is present but incomplete and inconsistent.

Observations:  
- Many job postings do not include salary information  
- Some entries have only minimum or maximum salary values  
- There may be inconsistencies such as unrealistic values or incorrect ranges  
- Multiple currencies may be present across records  

Salary normalization and handling missing values will be critical for salary analysis and modeling.

---

### Text Data Inspection

Text fields such as job title and job description require cleaning.

Observations:  
- Job descriptions may contain HTML tags and formatting noise  
- Some descriptions include boilerplate or repeated content  
- Job titles may include inconsistent formatting and seniority indicators  

Text cleaning will be necessary before applying NLP techniques.

---

### Final Data Quality Findings

Based on the exploration, the following key data quality issues were identified:

#### Missing Data  
- Salary fields are missing in a significant portion of the dataset  
- Some location and company fields are incomplete  

#### Consistency Issues  
- Employment type labels are inconsistent  
- Location fields are not standardized  
- Remote indicator is inconsistent or missing  

#### Duplication Issues  
- Duplicate job postings may exist  
- Some records share the same source identifiers  

#### Salary Issues  
- Salary data is incomplete and inconsistent  
- Multiple currencies are present  
- Salary ranges may not always be valid  

#### Text Quality Issues  
- Job descriptions contain HTML and noise  
- Job titles are not standardized  

These issues will guide the design of the data cleaning pipeline in the next step.
